Question 2. What is an interpreter in Python?

  > Interpretor are computer programs that  will convert source code or high-level language into intermediate code (Machine level Language).It is also called translator in programming terminology.
  Python interpretors process the code line by line, executing each instruction sequentially.

   How Python interpretor works -

   1. Lexical Analysis: Initially in source code analysis, Python get source and check for some Indentation rule and check for errors. if there are any errors Python stops its execution and make to modify to prevent errors. This process is called Lexical analysis in python which states that dividing the source code into list of individual tokens.

   2. Parsing: These tokens are then organized into a structure known as an Abstract Syntax Tree (AST), representing the grammatical structure of the code.

   3. Compilation to Bytecode: The AST is compiled into bytecode, a low-level, platform-independent representation of your code.

   4. Execution by Python Virtual Machine (PVM): The bytecode is executed by the PVM, which interprets the bytecode instructions and performs the corresponding operations on the host machine.

   Why Python Uses an Interpreter -

    Python is designed to be an interpreted language to enhance flexibility and ease of use. This design allows developers to write and execute code without the need for a separate compilation step, facilitating rapid development and testing. The line-by-line execution model also aids in debugging, as errors can be identified and addressed in isolation.

Question 6. Why are list mutable, but tuples are immutable?

  > The choice to make lists mutable and tuples immutable in Python is intentional and serves different purposes:

  Use Cases:

  > 1. Lists are designed for collections of items that may need to change, such as dynamic datasets.
  > 2. Tuples are intended for fixed collections of items, like coordinates or records, where the data should remain constant.

  Performance:

  > 1. Tuples can be more memory-efficient and faster than lists due to their immutability. This is because they don't require the overhead associated with dynamic resizing and other mutable operations.

  Hashability:

  > 1. Tuples can be used as keys in dictionaries because they are immutable and thus hashable. Lists cannot be used as dictionary keys because their mutability makes them unhashable.

Question 8. What is logical operators in python?

  > In Python, logical operators are used to combine conditional statements, allowing you to evaluate multiple expressions and control the flow of your program based on Boolean logic.
  > These operators return Boolean values (True or False) and are essential for decision-making in code.

  The Three Logical Operators in Python:

  > 1. and - Logical AND :  Returns True if both operands are True; otherwise, returns False.  Often used when multiple conditions must be met simultaneously.
  
  > 2. or - Logical OR : Returns True if at least one operand is True; returns False only if both operands are False. Useful when any one of multiple conditions being True is sufficient.

  > 3. not - Logical NOT : Reverses the Boolean value of the operand; not True becomes False, and not False becomes True. Employed to invert a condition's result.
  
  Examples are mentions below -

```python
# and -Logical AND
x = 10
print(x > 5 and x < 15)  # Output: True
```

```python
# or - Logical OR
x = 10
print (x < 5 or x > 8) # output: True
```

```python
# not - Logical NOT
x = 10
print (not( x > 5 and x < 15)) # result is True but not operator will invert it to false.
```

Question 9. What is Type Casting in Python?

  > Type casting, or type conversion, refers to converting a variable from one data type to another in Python.
  
  For example, if you have a variable containing a string number, such as "27", you may need to convert it to an integer before performing arithmetic operations. Otherwise, Python will interpret "27" as a string and add it to other strings instead of performing arithmetic.

Let's look at an example where we try to add a string number with an integer.

```python
# Type Conversion Example
a = "27" # str data type
b = 3    # int data type
direct_sum = a + b
print(direct_sum) # output will be Error (can only perform concatenate by using "," instead of "+")
```

```python
# Type Conversion Example
a = "27" # str data type
b = 3    # int data type
After_TypeCasting_Str = a + str(b)
print(After_TypeCasting_Str) # output will be 273

After_TypeCasting_Int = int(a) + b
print(After_TypeCasting_Int) # output will be 30
```

Question 11. What is the purpose of conditional statement in python?

  > The purpose of a conditional statement in Python is to allow the program to make decisions and execute different blocks of code based on whether a condition is True or False.

  >Conditional statements help control the flow of a program. They evaluate expressions using logical operators and run specific code depending on the outcome.

Common conditional statements in Python:

  - if: Executes a block of code if a condition is true.

  - elif: (else if) Tests another condition if the previous if or elif is false.

  - else: Executes a block of code if none of the previous conditions are true.

```python
# example of conditional statement if, else and elif .
number = 0

if number > 0:
    print ("The number is positive")
elif number < 0:
    print (" The number is negative ")
else:
    print ("The number is Zero")
```

Question 12. How does elif statement work?

  > The elif statement in Python stands for "else if" and is used to check additional conditions after an initial if statement. It runs only if the previous conditions are false.
  
  > In case multiplie elif statements python only check for the first True condition and give the result ignoring all elif statements after that.

How it works:

  - Python evaluates the if condition first.
  - If "if" is True, its block runs, and the rest (elif, else) is skipped.
  - If "if" is False, Python checks the elif condition.
  - If elif is True, its block runs and the rest is skipped.
  - If none of the if or elif conditions are True, the else block (if provided) runs.

```python
# Example of elif statement.

number = 50

if number >= 80:
    print ("Grade A")

elif number >= 60:
    print ("Grade B")

elif number >= 40:
    print ("Grade C")

else:
    print ("Fail")
```

Question 14. What is the difference between for and while loop?

  > The main difference between a for loop and a while loop in Python is how and when they check conditions and control repetition.

for loop

  - Used when you know in advance how many times you want to loop.
  - Often used to iterate over a sequence (like a list, tuple, string, or range).
  - Automatic Control no need to update variable manually.

while loop

  - Used when you don't know in advance how many times you'll loop.
  - Repeats as long as a condition is True.
  - Manual Control ( need to update variable manually ).

```python
# Example for loop
number = range(5)

for i in number:
    print (i)
```

```python
# Example while loop

row = 1

while row <=5:
    col = 1
    while col <= row:
      print ("*", end = " ")
      col = col + 1
    print()
    row = row + 1
```
